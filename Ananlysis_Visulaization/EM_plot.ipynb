```python
import numpy as np
import matplotlib.pyplot as plt

# Load RMSD data
data_md = np.loadtxt("rmsd.xvg", comments=["#", "@"])
data_xtal = np.loadtxt("rmsd_xtal.xvg", comments=["#", "@"])

# Extract columns
time_md = data_md[:, 0]
rmsd_md = data_md[:, 1]

time_xtal = data_xtal[:, 0]
rmsd_xtal = data_xtal[:, 1]

# Plot
plt.figure(figsize=(9, 5))

plt.plot(time_md, rmsd_md,
         color="royalblue", linewidth=1.5,
         label="RMSD vs. Initial MD Structure")

plt.plot(time_xtal, rmsd_xtal,
         color="tomato", linewidth=1.5,
         label="RMSD vs. Crystal Structure")

plt.xlabel("Time (ns)", fontsize=12)
plt.ylabel("Backbone RMSD (nm)", fontsize=12)
plt.title("Lysozyme Backbone RMSD - 10 ns MD Simulation", fontsize=13)

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

data = np.loadtxt("gyrate.xvg", comments=["#", "@"])

time = data[:, 0] / 1000  # ps to ns
rg = data[:, 1]

# Split trajectory into two halves
first_half = rg[time < 5]
second_half = rg[time >= 5]

# Compare averages
mean_first = np.mean(first_half)
mean_second = np.mean(second_half)

print(f"First 5 ns Mean Rg: {mean_first:.4f} nm")
print(f"Last 5 ns Mean Rg: {mean_second:.4f} nm")
print(f"Absolute Difference: {abs(mean_first - mean_second):.4f} nm")
```

```python
import numpy as np

data = np.loadtxt("gyrate.xvg", comments=["#", "@"])

time = data[:, 0] / 1000
rg = data[:, 1]

block_means = []

for i in range(10):
    mask = (time >= i) & (time < i + 1)
    block_means.append(np.mean(rg[mask]))

block_means = np.array(block_means)

for i, mean in enumerate(block_means):
    print(f"Block {i+1} ({i}-{i+1} ns): {mean:.4f} nm")

print(f"\nMean of block means: {np.mean(block_means):.4f} nm")
print(f"SD of block means: {np.std(block_means, ddof=1):.4f} nm")
```

```python
n_blocks = len(block_means)

block_se = np.std(block_means, ddof=1) / np.sqrt(n_blocks)

print(f"Mean Rg: {np.mean(block_means):.4f} nm")
print(f"Block SE: {block_se:.5f} nm")
```
